```python
# =========================================
# ZETTI SUPPLEMENTS
# Projeto Final - Lógica de Programação
# =========================================

produtos = [
    ["Whey Concentrado Chocolate", "900g", 119.90],
    ["Whey Concentrado Baunilha", "900g", 119.90],
    ["Whey Concentrado Morango", "900g", 119.90],

    ["Whey Isolado Chocolate", "900g", 179.90],
    ["Whey Isolado Baunilha", "900g", 179.90],
    ["Whey Isolado Cookies", "900g", 179.90],

    ["Creatina Monohidratada", "300g", 89.90],

    ["Cafeína 200mg", "120 cápsulas", 49.90],
    ["Cafeína 300mg", "60 cápsulas", 49.90],
    ["Cafeína 420mg", "60 cápsulas", 69.90],

    ["Pré-Treino Explosive 200mg", "300g", 109.90],
    ["Pré-Treino Nitro 300mg", "300g", 119.90],
    ["Pré-Treino Rage 420mg", "300g", 129.90],

    ["Ômega 3", "120 cápsulas", 59.90],
    ["BCAA", "300g", 69.90],

    ["Colágeno Tipo 1 e 3", "300g", 79.90],
    ["Colágeno Tipo 2", "60 cápsulas", 89.90]
]

carrinho = []

nome = ""
idade = 0
peso = 0
altura = 0

def calcular_imc(peso, altura):
    return peso / (altura ** 2)


def cadastro_cliente():
    global nome, idade, peso, altura

    print("=" * 40)
    print("  BEM-VINDO(A) AO ZETTI SUPPLEMENTS")
    print("=" * 40)

    nome = input("Nome: ")
    idade =  int(input("Idade: "))
    peso = float(input("Peso (kg): "))
    altura = float(input("Altura (m): "))

    imc = calcular_imc(peso, altura)

    print(f"\nBem-vindo, {nome}!")
    print(f"Seu IMC é {imc:.2f}")

    if imc < 18.5:
        print("Classificação: Abaixo do peso")
    elif imc < 25:
        print("Classificação: Peso normal")
    elif imc < 30:
        print("Classificação: Sobrepeso")
    else:
        print("Classificação: Obesidade")

def menu():
    print("\n" + "=" * 40)
    print("1 - Objetivos")
    print("2 - Catálogo Completo")
    print("3 - Ver Carrinho")
    print("4 - Finalizar Compra")
    print("0 - Sair")
    print("=" * 40)


def mostrar_objetivos():

    print("\nOBJETIVOS")

    print("1 - Ganho de Massa Muscular")
    print("2 - Emagrecimento")
    print("3 - Performance")
    print("4 - Saúde")

    opcao = int(input("Escolha: "))

    print()

    if opcao == 1:
        print("Recomendação:")
        print("- Whey Concentrado")
        print("- Creatina")
        print("- BCAA")

    elif opcao == 2:
        print("Recomendação:")
        print("- Whey Concentrado/Isolado")
        print("- Cafeína")
        print("- Pré-Treino")

    elif opcao == 3:
        print("Recomendação:")
        print("- Creatina")
        print("- Pré-Treino")
        print("- BCAA")

    elif opcao == 4:
        print("Recomendação:")
        print("- Ômega 3")
        print("- Colágeno")
        print("- Whey Isolado")


def mostrar_catalogo():

    print("\nCATÁLOGO DE PRODUTOS")

    for i in range(len(produtos)):
        print(f"{i + 1} - {produtos[i][0]}")
        print(f"    Quantidade: {produtos[i][1]}")
        print(f"    Preço: R${produtos[i][2]:.2f}")

    adicionar = input("\nDeseja adicionar ao carrinho? (S/N): ")

    if adicionar.upper() == "S":

        codigo = int(input("Número do produto: "))

        if codigo >= 1 and codigo <= len(produtos):

            produto = produtos[codigo - 1]

            nome_produto = produto[0]

            if idade < 16 and ("Cafeína" in nome_produto or "Pré-Treino" in nome_produto):
                print("\n**ATENÇÃO!**")
                print("Venda não permitida para menores de 16 anos!")

            else:
                carrinho.append(produto)
                print("Produto adicionado ao carrinho!")

        else:
            print("Produto inválido.")


def ver_carrinho():

    if len(carrinho) == 0:
        print("\nCarrinho vazio.")
        return

    total = 0

    print("\nSEU CARRINHO")

    for i in range(len(carrinho)):
        print(f"{i + 1} - {carrinho[i][0]} - R${carrinho[i][2]:.2f}")
        total += carrinho[i][2]

    print(f"\nTotal parcial: R${total:.2f}")


def finalizar_compra():

    if len(carrinho) == 0:
        print("\nCarrinho vazio.")
        return

    subtotal = 0

    for item in carrinho:
        subtotal += item[2]

    desconto = 0

    if subtotal >= 500:
        desconto = subtotal * 0.10

    elif subtotal >= 300:
        desconto = subtotal * 0.05

    total = subtotal - desconto

    print("\n" + "*" * 40)
    print("           NOTA FISCAL")
    print("*" * 40)

    print(f"Cliente: {nome}")

    print()

    for item in carrinho:
        print(f"{item[0]}......R${item[2]:.2f}")

    print("\n" + "-" * 40)

    print(f"Subtotal:......R${subtotal:.2f}")
    print(f"Desconto:......R${desconto:.2f}")
    print(f"Total da compra:......R${total:.2f}")

    print("\nObrigado pela compra!")

    print("=" * 40)


cadastro_cliente()

opcao = -1

while opcao != 0:

    menu()

    opcao = int(input("Escolha uma opção: "))

    if opcao == 1:
        mostrar_objetivos()

    elif opcao == 2:
        mostrar_catalogo()

    elif opcao == 3:
        ver_carrinho()

    elif opcao == 4:
        finalizar_compra()

    elif opcao == 0:
        print("\nObrigado por visitar a Zetti Supplements!")

    else:
        print("Opção inválida.")
```
